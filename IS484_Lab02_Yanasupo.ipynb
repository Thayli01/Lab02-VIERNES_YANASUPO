## Mapa de la ciudad

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
    }
print(mapa)
```

## Algoritmo BFS

```python
# Representación del mapa como diccionario de adyacencia
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

# Algoritmo de Búsqueda en Amplitud (BFS)
def bfs(grafo, inicio, objetivo):
    cola = [[inicio]]  # Inicializa la cola FIFO con el camino inicial
    visitados = set()  # Conjunto para rastrear los nodos ya evaluados

    while cola:  # Procesa mientras queden caminos por explorar
        camino = cola.pop(0)  # Extrae el PRIMER camino (comportamiento FIFO)
        nodo_actual = camino[-1]  # Extrae el último nodo del camino actual

        if nodo_actual == objetivo:  # Evalúa si se alcanzó la meta
            return camino  # Devuelve el camino más corto encontrado

        if nodo_actual not in visitados:  # Evita evaluar nodos repetidos
            visitados.add(nodo_actual)  # Marca el nodo actual como visitado

            for vecino in grafo[nodo_actual]:  # Recorre las conexiones del nodo
                nuevo_camino = camino + [vecino]  # Construye la nueva trayectoria
                cola.append(nuevo_camino)  # Encola la nueva ruta al final

    return None  # Devuelve None si no existe ruta hacia el objetivo


# Prueba de ejecución del algoritmo
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (BFS):", resultado)
```

## Planteamiento del problema
Se quiere encontrar, dentro de un conjunto de observaciones ya conocidas, cuál es la más parecida a

una observación nueva  el mismo principio que usará el algoritmo KNN a partir de la Semana 7.

```python
# Algoritmo DFS completado
def dfs(grafo, inicio, objetivo): 
    pila = [[inicio]]
    visitados = set()
    while pila: # con el bulce while recorremos la pila mientras tenga elementos 
        camino = pila.pop() # se saca el ultimo 
        nodo_actual = camino[-1] # se obtiene el ultimo nodo del camino 
        if nodo_actual == objetivo: # si el nodo actual es el estrado objetivo retornamos el camino 
            return camino
        if nodo_actual not in visitados:# evaluams si el nodo ya fue visitado 
            visitados.add(nodo_actual) # si no fue visitado agregar 
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)
    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

BFS permite encontrar la ruta con menor número de pasos (['Almacen', 'Centro', 'ZonaNorte']), mientras que DFS explora hasta el fondo usando una estructura LIFO (pila) y no asegura el camino más corto.

## Encontrar la observación más cercana

Utlizando elm data set de temperatura y humedad, se agrega una nueva observacion. 
Calcular entre la nueva observacion con cada una de las existentes para identificar la ams cercana

```python

import numpy as np

obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

dataset = np.array([obs1, obs2, obs3, obs4]) #dataset 
nueva_observacion = np.array([20.5, 48])
distancias = []

for i, obs in enumerate(dataset):
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observacion {i}: {d:.4f}")

indice_mas_cercano = np.argmin(distancias)
print("La observacion mas parecida es la numero:", indice_mas_cercano)
```

## Refexión

Utilidad: La distancia euclidiana
$ d(\vec{a},\vec{b}) = \sqrt{\sum_{i=1}^{n}(a_i-b_i)^2} $
cuantifica la separacion entre vectores en un espacio multidimensional. 
Cuanto menor sea la distancia, mayor será la similitud entre sus propiedades.

Limitación: La distancia euclidiana es sensible a las diferencias de escala. Si una característica se mide en miles, como los ingresos en soles, y otra en unidades pequeñas, como la edad en años, la variable de mayor magnitud puede dominar el cálculo. Por ello, puede ser necesario realizar previamente una estandarización o normalización de los datos.
